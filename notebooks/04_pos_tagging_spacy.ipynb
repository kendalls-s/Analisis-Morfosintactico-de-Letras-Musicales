# Importación de Librerías y Carga del Modelo spaCy


```python
import sys
import os
import warnings
import spacy
import pandas as pd
from pathlib import Path
warnings.filterwarnings('ignore')
sys.path.append(os.path.abspath('../../../..'))
from src.data.mongo_storage import _get_default_collection
from src.data.preprocessor import token_spacy
from src.pos_tagging.Taggin_spacy import apply_pos_tagging_spacy

print('✓ Librerías importadas correctamente')
```

# Carga del Dataset


```python
# Cargar canciones desde MongoDB
col = _get_default_collection()
docs = list(col.find({}, {
    '_id': 1, 'Song': 1, 'Artist': 1, 'Genre': 1,
    'Song year': 1, 'Lyrics': 1, 'Language': 1
}))

df = pd.DataFrame(docs)
print(f'✓ {len(df)} canciones cargadas desde MongoDB')
df.head()
```

# Tokenización

```python
df = token_spacy(df)
print('✓ Tokenización con spaCy completada')
df.head()
```

# POS Taggin



```python
df = apply_pos_tagging_spacy(df)
print('✓ POS tagging con spaCy completado y guardado en MongoDB')
df.head()
```
